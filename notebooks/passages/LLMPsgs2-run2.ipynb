```python
slices = slice_txt(txt,window=500)
len(slices)
```

```python
with open('system_prompt.char_desc.md','r') as f:
    system_prompt = f.read()

# print(system_prompt)
```

```python
def iter_slices(slices,n=10):
    slices_id = list(enumerate(slices))
    for i in range(0,len(slices_id),n):
        this_slices_id = slices_id[i:i+n]
        sep = '\n\n' + ('-' * 100) + '\n\n'


        user_prompt = sep.join(
            f'#########\nslice_{i:03d}\n#########\n\n{s}'
            for i,s in this_slices_id
        )

        ids = [i for i,s in this_slices_id]
        yield tuple(ids),user_prompt
```

```python
import time

# print(user_prompt)
for ids,user_prompt in tqdm(list(iter_slices(slices,10))):
    response = generate_json(
        user_prompt,
        system_prompt=system_prompt,
        model=DEFAULT_MODEL,
        cache_key={'text_id':text_id, 'slice_ids':ids},
        verbose=False
    )
    time.sleep(0.1)
    # break
```
